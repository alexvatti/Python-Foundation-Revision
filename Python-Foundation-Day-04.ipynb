```python
# Print 1 to 5 using recursion.
def print_number(n):
    if n > 5: # end case # not satified # exit # return
        return
    print(n)
    print_number(n+1)
print_number(1) # initial val
```

```python
# 5 to 1 
n = 5
while n >= 1: # satisfied -  enter loop
    print(n)
    n = n - 1
```

```python
# Print 5 to 1 using recursion.
def print_number(n):
    if n <= 0: # reverse # exit
        return
    print(n)
    print_number(n-1)
print_number(5)
```

```python
# Print 1 to n(5). 1,2,3,4,5,6,7,8,9,20
def print_number(initial_val,target):
    if initial_val  > target:
        return
    print(initial_val)
    print_number(initial_val+1,target)
print_number(5,7)
```

```python
# Print n to 1.
def print_number(initial_val,target):
    if initial_val < target:
        return
    print(initial_val)
    print_number(initial_val-1,target)
print_number(5,1)
```

```python
# Print even numbers from 2 to 10.
def even_numbers(n,m):
    if n > m:
        return
    if n%2==0:
        print(n)
    even_numbers(n+1,m)
even_numbers(2,10)
```

```python
# Print even numbers from 2 to 10.
def odd_numbers(n,m):
    if n > m:
        return
    if n%2!=0:
        print(n)
    odd_numbers(n+1,m)
odd_numbers(2,10)
```

## Level 2 — Easy
1. Find the sum of 1 to n.
2. Find the factorial of n.
3. Count the number of digits in a number.
4. Find the sum of digits of a number.
5. Calculate a raised to power n.

```python
# Find the sum of 1 to n.
def sum_recursion(initial,target):
    if initial > target:
        return 0
    return initial + sum_recursion(initial+1, target)
print(sum_recursion(1,5))
```

```python
# Find the factorial of n.
def fact(n):
    if n == 0:
        return 1
    return n * fact(n-1)
print(fact(5))
    
```

```python
# Count the number of digits in a number.
def count_digits(n): # 1234
    if n==0:
        return 0
    return 1 + count_digits(n//10) # 1+ count_digits(123) 10 
print(count_digits(1234))
```

```python
# Find the sum of digits of a number.
def sum_digits(n):
    if n==0:
        return 0
    return n%10 + sum_digits(n//10)
print(sum_digits(1))
```

```python
# Calculate a raised to power n.
10*10*10
def raise_power(base,n):
    if n<1:
        return 1
    return base*raise_power(base,n-1)
print(raise_power(10,2))
```

# Level 3 — Medium
1. Reverse a string using recursion.
2. Check whether a string is a palindrome.
3. Find the maximum element in a list.
4. Find the sum of elements in a list.
5. Search for an element in a list.

```python
# Reverse a string using recursion.
def rev_string(str):
    if len(str) == 0:
        return ""
    return str[-1] + rev_string(str[:-1])
print(rev_string("hello"))
```

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False

    return is_palindrome(s[1:-1])


print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

```python
# Find the maximum element in a list.
def find_max(numbers):
    if len(numbers) == 1:
        return numbers[0]

    max_remaining = find_max(numbers[1:])

    if numbers[0] > max_remaining:
        return numbers[0]
    else:
        return max_remaining


numbers = [10, 25, 7, 40, 15]

print(find_max(numbers))
    
```

```python
def find_min(numbers):
    if len(numbers) == 1:
        return numbers[0]

    min_remaining = find_min(numbers[1:])

    if numbers[0] < min_remaining:
        return numbers[0]
    else:
        return min_remaining


numbers = [10, 25, 7, 40, 15]

print(find_min(numbers))
```

```python
# Find the sum of elements in a list.
def sum_list(lst):
    if len(lst)==0:
        return 0

    return lst[0] + sum_list(lst[1:])
print(sum_list([1,2,3,4]))
```

```python
# Find the product of elements in a list.
def prod_list(lst):
    if len(lst)==0:
        return 1

    return lst[0] * prod_list(lst[1:])
print(prod_list([1,2,3,4]))
```

```python
# Search for an element in a list.
def search_ele_list(lst,num):
    if len(lst)==0:
        return False
    if num == lst[0]:
        return True
    return search_ele_list(lst[1:],num)
print(search_ele_list([1,2,3,4],5))
```

```python
def fibonacci(n):
    if n<=1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
print(fibonacci(5))   
```

```python
def binary_search(num , lst):

    if len(lst) == 0:
        return False
        
    low_index = 0
    high_index = len(lst)
    mid_index = (low_index + high_index)//2

    if lst[mid_index]==num:
        return True
    elif lst[mid_index]< num:
        low_index = mid_index +1
    else:
        high_index = mid_index
    return binary_search(num, lst[low_index:high_index])
print(binary_search(70, [10, 20, 30, 40, 50, 60, 70]))   
```

```python
# a*b = a if b==1
# a*b = a + sum(a,b-1)

def prod_sum(a,b):
    if b==1:
        return a
    else:
        return a + prod_sum(a,b-1)
print(prod_sum(6,3))
```

```python
def average_list(lst):
    if len(lst) == 0:
        return 0, 0

    total, count = average_list(lst[1:])

    return lst[0] + total, count + 1


total, count = average_list([10, 20, 30, 40])

print(total / count)
```

```python
def search_ele_list(lst,num):
    if len(lst)==0:
        return 0
    if num == lst[0]:
        return 1 + search_ele_list(lst[1:],num)
    else: 
        return search_ele_list(lst[1:],num)
print(search_ele_list([1,2,2,3,4],2))
```

```python
def search_ele_list_first(lst,num,index):
    if len(lst)==0:
        return -1
    if num == lst[0]:
        return index
    else: 
        return search_ele_list_first(lst[1:],num,index+1)
print(search_ele_list_first([1,2,2,3,4],100,0))
```

```python
def search_ele_list_last(lst, num, index):
    if len(lst) == 0:
        return -1

    result = search_ele_list_last(lst[1:], num, index + 1)

    if result != -1:
        return result

    if num == lst[0]:
        return index

    return -1


print(search_ele_list_last([1, 2, 2, 3, 4], 2, 0))
```

```python
def is_sorted(lst):
    if len(lst) <= 1:
        return True

    if lst[0] > lst[1]:
        return False

    return is_sorted(lst[1:])
print(is_sorted([1, 2,2, 3, 4]))
```

````python
def permutations(s):

    for i in range(len(s)):# len of string 3
        current = s[i]
        remaining = s[:i] + s[i + 1:]
        print(f"current={current},remaining={remaining}")
permutations("BCD") # 4 *3 *2 = P
````

````python
def permutations(s):
    if len(s) <= 1:
        return [s]

    result = []

    for i in range(len(s)):# len of string 3
        current = s[i]
        remaining = s[:i] + s[i + 1:]
        print(f"current={current},remaining={remaining}")

        for p in permutations(remaining): # number of times
            print(current,p)
            result.append(p +current )  # A, BC -> BC - B ,C ->ABC
            

    return result


print("Permutations:")
print(permutations("ABC"))
````

````python
def numbers():
    yield 10
    yield 20
    yield 30


for num in numbers():
    print(num)
````

````python
def numbers():
    yield 10
    yield 20
    yield 30


gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
````

````python
def normal_function():
    return 10


def generator_function():
    yield 10


print(normal_function())

gen = generator_function()

print(next(gen))
````

````python
def count_numbers(n):
    for i in range(1, n + 1):
        yield i


for num in count_numbers(5):
    print(num)
````

````python
def even_numbers(n):
    for i in range(2, n + 1, 2):
        yield i


for num in even_numbers(10):
    print(num)
````

````python
def squares(n):
    for i in range(1, n + 1):
        yield i * i


for value in squares(5):
    print(value)
````

````python
numbers = (x * x for x in range(1, 6))

for num in numbers:
    print(num)
````

````python
def decorator(function):

    def wrapper():
        print("Before function")

        function()

        print("After function")

    return wrapper


def greet():
    print("Hello")


greet = decorator(greet)

greet()
````

````python
def decorator(function):

    def wrapper():
        print("Before function")

        function()

        print("After function")

    return wrapper


@decorator
def greet():
    print("Hello")


greet()
````

````python
def decorator(function):

    def wrapper(name):
        print("Before function")

        function(name)

        print("After function")

    return wrapper


@decorator
def greet(name):
    print("Hello", name)


greet("Alex")
````

````python
def decorator(function):

    def wrapper(*args):
        print("Before function")

        function(*args)

        print("After function")

    return wrapper


@decorator
def add(a, b):
    print(a + b)


add(10, 20)
````

````python
def decorator(function):

    def wrapper(*args, **kwargs):
        print("Before function")

        result = function(*args, **kwargs)

        print("After function")

        return result

    return wrapper


@decorator
def greet(name, age):
    print(name, age)


greet("Alex", 45)
````

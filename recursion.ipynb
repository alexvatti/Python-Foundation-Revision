# Recursion
1. loop method (for / while loop)
2. own stack method
3. recusion - os stack method

```python
# method loop method
def recursion_sum_digits(n):
    s=0
    while (n!=0):
        d = n % 10
        s = s +d
        n = n // 10
    return s
print(recursion_sum_digits(1234))
```

```python
# method -own stack
def recursion_sum_digits(n, stack=None):
    if stack is None:
        stack = []

    if n == 0:
        s = 0
        while stack:
            val = stack.pop()
            s += val
        return s
    else:
        d = n % 10
        stack.append(d) # push
        n = n // 10
        return recursion_sum_digits(n, stack)


print(recursion_sum_digits(1234))  # 10
```

```python
# full recursion
def recursion_sum_digits(n):
    if n == 0:
        return 0
    else:
        d = n % 10
        n = n // 10
        return d +  recursion_sum_digits(n)  # Recursive call first


print(recursion_sum_digits(1234))  # 10
```

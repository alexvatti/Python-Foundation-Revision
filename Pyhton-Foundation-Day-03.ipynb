```python
import math
v1 = [1,2,3,4,5,6,7]
v2 = [1,3,4,5,6,7,8]

#cos = A.B / MAG(A)x MAG(B)

def cos_theta(v1,v2):
    a_dot_b = 0
    mag_a = 0
    mag_b = 0
    for pair in zip(v1,v2):
        a_dot_b = a_dot_b + pair[0]*pair[1]
    for val in v1:
        mag_a = mag_a+ math.sqrt(val**2)

    for val in v2:
        mag_b = mag_b+ math.sqrt(val**2)
  
    cos_value = a_dot_b / (mag_a*mag_b)
    print(cos_value)

cos_theta(v1,v2)
```

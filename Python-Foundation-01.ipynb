```python
a = [10, 20, 30]

b = a.copy()

b[0] = 100

print(a)
print(b)
```

```python
a = [[10, 20], [30, 40]]

b = a.copy()

b[0][0] = 100
print(id(a))
print(id(b))
print(a)
print(b)
```

```python
import copy
a = [[10, 20], [30, 40]]

b = copy.deepcopy(a)

b[0][0] = 100
print(id(a))
print(id(b))
print(a)
print(b)
```

copy.copy()       → shallow copy
copy.deepcopy()   → completely separate nested objects

# List vs Tuple vs Set vs Dictionary

## list 
1. Ordered (indexed) and changeable(mutable).

## tuple
1. Ordered (indexed) but cannot be changed(immutable).

## set
1. Stores unique values.(mutable)

## dictionary
1. Stores key → value.(mutable) (index-using key)

List       → ordered collection

Tuple      → fixed collection

Set        → unique values

Dictionary → key-value data

```python
def add(*args):
    print(args)

add(10, 20, 30,40)
```

```python
def student(**kwargs):
    print(kwargs)

student(name="Alex", age=45)
```
